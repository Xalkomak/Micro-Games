```python
import random
```

```python
CELL_STATES = ("UNOPENED", "OPENED", "FLAGGED", "MINE") # Final state only used when the player uncovers a mine
TOLERANCE = 3 # To be used in case of num_mines being more than the product of rows and columns
```

```python
def print_board(board_display: list, minefield: list, show_mines: bool = False):
  rows = len(board_display)
  columns = len(board_display[0])

  # Emojis for display
  UNOPENED_EMOJI = "⬛"
  FLAGGED_EMOJI = "🚩"
  MINE_EMOJI = "💣"
  EMPTY_OPENED_EMOJI = "O" # Placeholder, will be numbers later

  # Print column headers
  print("  " + "  ".join(str(i) for i in range(columns)))
  print(" " + "---" * columns)

  for r in range(rows):
    print(f"{r:02d}|", end=" ") # Print row header
    for c in range(columns):
      if board_display[r][c] == CELL_STATES[0]: # UNOPENED
        if show_mines and minefield[r][c]:
          print(MINE_EMOJI, end="") # Show mine in debug mode
        else:
          print(UNOPENED_EMOJI, end="") # Unopened cell
      elif board_display[r][c] == CELL_STATES[1]: # OPENED
        if minefield[r][c]: # If opened cell is a mine, it's a game over state
            print(MINE_EMOJI, end="")
        else:
            print(EMPTY_OPENED_EMOJI, end=" ") # Placeholder for numbers
      elif board_display[r][c] == CELL_STATES[2]: # FLAGGED
        print(FLAGGED_EMOJI, end="") # Flagged cell
      else: # This case should ideally not be hit with current CELL_STATES usage unless it's an 'opened mine'
        print("?", end=" ") # Fallback for unexpected state
    print()
```

```python
def field_setup(num_mines: int, rows: int = 8, columns: int = 8) -> tuple[list, list]:
  minefield = [[False for _ in range(columns)] for _ in range(rows)] # Setup the minefield
  cell_states = [[CELL_STATES[0] for _ in range(columns)] for _ in range(rows)] # Copy the same setup from above to mark the state of all cells
  num_mines = min(num_mines, round((rows * columns)/(TOLERANCE))) # Ensure the number of mines is not greater than the total number of cells

  mine_count = 0
  while mine_count < num_mines:
    row = random.randint(0, rows - 1)
    column = random.randint(0, columns - 1)
    if minefield[row][column] == False:
      minefield[row][column] = True
      mine_count += 1

  return minefield, cell_states
```

```python
minefield_data, board_display_states = field_setup(100)

print("Initial Player's View:")
print_board(board_display_states, minefield_data)

print("\nMinefield (for Debugging/Revealing):")
print_board(board_display_states, minefield_data, show_mines=True)
```
