```python
import random
import time
from IPython.display import clear_output
```

```python
def check_running():
  return (p1_score > loss_score or p1_score < goal_score) or (p2_score > loss_score or p2_score < goal_score)
```

```python
def roll_green_dice():
  return random.randint(1, 6)

def roll_red_dice():
  return random.randint(1, 6)
```

```python
def event3():
  print("Event 3 triggered! You get a second roll.")
  game_loop(True)
```

```python
def event4():
  print("Event 4 triggered! Choose your move.")
  print("1. Roll a dice to add to your score")
  print("2. Roll a dice to subtract from your opponent's score")

  global p1_score, p2_score, p1_turn

  choice = 0
  while choice not in [1, 2]:
    user_input = input("Enter your choice (1 or 2): ")
    try:
      choice = int(user_input)
      if choice not in [1, 2]:
        print("Invalid choice. Please enter 1 or 2.")
    except ValueError:
      print("Invalid input. Please enter a number (1 or 2).")
      choice = 0 # Reset choice to keep loop running

  roll = random.randint(1, 6)
  print(f"You rolled a {roll}.")

  if choice == 1:
    if p1_turn:
      p1_score += roll
    else:
      p2_score += roll
  else:
    if p1_turn:
      p2_score -= roll
    else:
      p1_score -= roll
```

```python
def event7(green, red):
  print("Event 7 triggered!")
  global p1_score, p2_score

  additional_score = (green - red) * 3 # Potentially dangerous formula
  print(f"Additional score: ({green} - {red}) * 3 = {additional_score}")
  if p1_turn:
    p1_score += additional_score
  else:
    p2_score += additional_score
```

```python
def print_scores():
  print(f"Player 1's score: {p1_score}")
  print(f"Player 2's score: {p2_score}")
```

```python
'''
Game loop procedure:
1. Player rolls both dice.
2. Check for Rule of Doubles.
3. Update player scores. Use Rule of Doubles to double scores if applicable.
4. Check for Event-X conditions and execute applicabel ones.
'''

def game_loop(skip_events = False):
    global p1_score, p2_score, p1_turn, current_round

    if not skip_events:
      print(f"Round {current_round}")
      current_round += 1
      print_scores()
      print()

      if p1_turn:
        print("Player 1's turn")
      else:
        print("Player 2's turn")
      input("Press 'ENTER' to roll the dice.")

    green = roll_green_dice()
    red = roll_red_dice()
    print()

    if not skip_events:
      print("Green Dice: ", green)
      print("Red Dice: ", red)
      if green == red: # Rule of Doubles
        print("Rule of Doubles triggered! Double scores.")
        green *= 2
        red *= 2
        print("Green Dice: ", green)
        print("Red Dice: ", red)
    else:
      print("Second Roll for Event-3:")
      print("Green Dice: ", green)
      print("Red Dice: ", red)

    # Update player scores
    if p1_turn:
      p1_score += green
      p2_score -= red
    else:
      p2_score += green
      p1_score -= red

    print()
    # Event-X checks
    if not skip_events:

      # RoD, if triggered, doubles the values of the dice.
      # This step halves the values, bringing them back the original values.
      # We know this works because RoD has to trigger earlier above
      # for this to trigger, or else it gets skipped.
      if green == red:
        green /= 2
        green = int(green)
        red /= 2
        red = int(red)

      if green + red == 3:
        event3()
      if green + red == 4:
        event4()
      if green + red == 7:
        event7(green, red)

    print_scores()
    input("\nPress 'ENTER' to begin the next round.")
    clear_output()
    p1_turn = not p1_turn
```

```python
current_round = 1

# Initial scores, can be changed
p1_score = 50
p2_score = 50
goal_score = 100
loss_score = 0

p1_turn = True # Oscillates between the two players

# ISSUE: Google Colab stops properly running after a certin number of turns

is_game_running = True
while is_game_running:
  is_game_running = check_running()
  game_loop()
```
